```python
from torchvision.datasets import Food101

ds = Food101(root="data", split="train", download=True)
print("Images:", len(ds))
print("Classes:", len(ds.classes))
print(ds.classes)
```

```python
import random
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
for i, idx in enumerate(random.sample(range(len(ds)), 8)):
    img, label = ds[idx]
    plt.subplot(2, 4, i + 1)
    plt.imshow(img)
    plt.title(ds.classes[label])
    plt.axis("off")
plt.show()
```

```python
import matplotlib.pyplot as plt
import random

plt.figure(figsize=(12, 6))
for i in range(8):
    idx = random.randint(0, len(ds) - 1)
    img, label = ds[idx]
    plt.subplot(2, 4, i + 1)
    plt.imshow(img)
    plt.title(ds.classes[label])
    plt.axis("off")
plt.tight_layout()
plt.show()
```
